```python
#Program that finds maximum among three numbers
a=int(input('Enter first number'))
b=int(input('Enter second number'))
c=int(input('Enter third number'))
if(a>b and a>c):
  print(a,' is greater among three numbers')
elif(b>c):
  print(b,'is greater among three numbers')
else:
  print(c,'is greater among three numbers')


```

```python
#Program to find a leap year or not
year=int(input('Enter a year'))
if(year%4==0 and year%100!=0) or (year%400==0):
  print(f'{year} is leap year')
else:
  print(f'{year} is not leap year')
```

```python
#Program on match case
day=int(input('Enter'))
match day:
  case 1:
    print('Sunday')
  case 2:
    print('Monday')
  case 3:
    print('Tuesday')
  case 4:
    print('Wednesday')
  case 5:
    print('Thusrday')
  case 6:
    print('Friday')
  case 7:
    print('Saturday')
```

```python
#Adding and multiplying two complex numbers
i=complex(input('Enter first number'))
j=complex(input('Enter second number'))
sum=i+j
mul=i*j
print(f'sum is {sum}')
print(f'mul is {mul}')
```

```python
#Swap two variable without using temporary variable
x=int(input('Enter number'))
y=int(input('Enter another number'))
print(x,y)
x,y = y,x
print(x,y)
```

```python
#Print 1 to n even numbers
c=int(input('Enter the number'))
while c>=1:
  if c%2==0:
    print(c,end=' ')
  c-=1
```

```python
#sum of digits of a given number
e=int(input('enter number'))
sum=0
rem=0
while e!=0:
  rem=e%10
  sum=sum+rem
  e=e//10
print(sum)
```

```python
#reverse
f=int(input('enter number'))
sum=0
rem=0
while f!=0:
  rem=f%10
  sum=sum*10+rem
  f=f//10
print(sum)
```

```python
#prime or not
g=int(input())
for i in range(2,g+1):
  if g%i==0:
    print('not a prime number')
    break
  else:
    print('a prime number')
```

```python
#print prime numbers between two numbers
h = int(input())
k = int(input())
for i in range(h,k+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=' ')
```
